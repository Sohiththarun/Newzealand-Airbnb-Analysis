```python
import pandas as pd

files = {
    "October 2025": "October_2025.csv",
    "November 2025": "November_2025.csv",
    "December 2025": "December_2025.csv",
    "January 2026": "January_2026.csv",
    "February 2026": "February_2026.csv",
    "March 2026": "March_2026.csv",
    "April 2026": "April_2026.csv",
    "May 2026": "May_2026.csv",
    "June 2026": "June 2026.csv"
}
```

```python
all_christchurch = []

for month, file in files.items():

    # Load dataset
    data = pd.read_csv(file)

    # Filter Christchurch
    data = data[
        data["neighbourhood_group"] == "Christchurch City"
    ].copy()

    # Add month and year
    data["month_year"] = month

    # Convert date
    data["last_review"] = pd.to_datetime(
        data["last_review"],
        errors="coerce"
    )

    # Store processed dataset
    all_christchurch.append(data)
```

Concatenate the datasets

```python
christchurch_all = pd.concat(
    all_christchurch,
    ignore_index=True
)
```

```python
import pandas as pd

files = {
    "October 2025": "October_2025.csv",
    "November 2025": "November_2025.csv",
    "December 2025": "December_2025.csv",
    "January 2026": "January_2026.csv",
    "February 2026": "February_2026.csv",
    "March 2026": "March_2026.csv",
    "April 2026": "April_2026.csv",
    "May 2026": "May_2026.csv",
    "June 2026": "June 2026.csv"
}

all_data = []

for month, file in files.items():

    data = pd.read_csv(file)

    # Christchurch only
    data = data[
        data["neighbourhood_group"] == "Christchurch City"
    ].copy()

    # Add the month of the dataset
    data["month_year"] = month

    all_data.append(data)

# Combine all months
df = pd.concat(all_data, ignore_index=True)
```

```python
print(df.columns)
```

```python
print(df["month_year"].value_counts())
```

```python
import matplotlib.pyplot as plt

month_order = [
    "October 2025",
    "November 2025",
    "December 2025",
    "January 2026",
    "February 2026",
    "March 2026",
    "April 2026",
    "May 2026",
    "June 2026"
]

df["month_year"] = pd.Categorical(
    df["month_year"],
    categories=month_order,
    ordered=True
)

monthly_price = (
    df.groupby("month_year", observed=True)["price"]
      .mean()
      .reset_index()
)

plt.figure(figsize=(12, 8))

plt.plot(
    monthly_price["month_year"],
    monthly_price["price"],
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Average Price (NZD)")
plt.title("Average Airbnb Price in Christchurch: October 2025 – June 2026")

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(df["days_since_latest_review"].dropna(), bins=50)

plt.xlabel("Days Since Latest Review")
plt.ylabel("Number of Listings")
plt.title("Distribution of Days Since Latest Review")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df.loc[df["price"] <= 1000, "price"].dropna(),
    bins=30
)

plt.xlabel("Price (NZD)")
plt.ylabel("Frequency")
plt.title("Christchurch Airbnb Price Distribution (Prices ≤ $1000)")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df["publication_date"] = pd.NaT

df.loc[
    df["month_year"] == pd.Timestamp("2025-10-01"),
    "publication_date"
] = pd.Timestamp("2025-10-05")
```

```python
df["last_review"] = pd.to_datetime(
    df["last_review"],
    errors="coerce"
)
```

```python
df["days_since_latest_review"] = (
    df["publication_date"] - df["last_review"]
).dt.days
```

```python
df[
    ["publication_date", "last_review",
     "days_since_latest_review"]
].head(20)
```

```python
average_days = df["days_since_latest_review"].mean()

print("Average days since latest review:",
      round(average_days, 2))
```

```python
import matplotlib.pyplot as plt

# Remove missing review differences
review_days = df["days_since_latest_review"].dropna()

# Create bins so the bars are readable
bins = [0, 50, 100, 150, 200, 250, 300, 350, 400, 450, 500, 550, 600, 650, 700, 750, 800, 850, 900, 950, 1000]

plt.figure(figsize=(16, 6))

plt.hist(
    review_days,
    bins=bins
)

plt.xlabel("Days Since Latest Review")
plt.ylabel("Number of Properties")
plt.title(
    "Days Since Latest Review - Christchurch Airbnb Properties (October 2025)"
)

plt.xticks(bins, rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
