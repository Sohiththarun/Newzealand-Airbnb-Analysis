Load your Deliverable 3 output

```python
import pandas as pd

christchurch_all = pd.read_csv(
    "output/Christchurch_listings_Oct2025_Jun2026.csv"
)

christchurch_all.shape
```

Handle missing values ( In deliverable I found these are all missing values price, last_review, and reviews_per_month)

```python
before_price = len(christchurch_all)

christchurch_all = christchurch_all.dropna(
    subset=["price"]
).copy()

after_price = len(christchurch_all)

print("Rows removed:", before_price - after_price)
print("Rows remaining:", after_price)
```

```python
christchurch_all["last_review"] = pd.to_datetime(
    christchurch_all["last_review"],
    errors="coerce"
)
```

Check duplicates

```python
print("Invalid prices:",
      (christchurch_all["price"] <= 0).sum())

print("Invalid minimum nights:",
      (christchurch_all["minimum_nights"] <= 0).sum())

print("Invalid availability:",
      ((christchurch_all["availability_365"] < 0) |
       (christchurch_all["availability_365"] > 365)).sum())
```

The final cleaned listing dataset

```python
christchurch_clean = christchurch_all.drop(
    columns=["name", "host_name", "license"]
).copy()
```

Save the cleaned Christchurch dataset

```python
christchurch_clean.to_csv(
    "cleaned_christchurch_listings.csv",
    index=False
)
```

Document our cleaning decisions:

The Christchurch listing dataset was cleaned --> prepare to rental bond data.

The name column was removed because it identifies individual listings and is not required for comparing rental prices or property availability.

The host_name column was removed --> it is not relevant to main question

The license column was removed --> no useful--> non missing information

Rows with missing price values were removed because rental price is a key variable for the planned analysis. A missing price cannot be reliably used in price comparisons.

Rows with missing last_review and reviews_per_month were retained because these variables are not central to the main rental price and availability analysis. Missing review information can occur naturally for listings with little or no review history.

Duplicate rows were checked and removed if present.

Latitude and longitude were retained because the assignment requires them and they may be useful for future geographic analysis.

The month_year variable was retained because the dataset needs to be compared with the rental bond data over the same time period.

Document the consequences

```python
initial_rows = len(christchurch_all)



final_rows = len(christchurch_clean)

print("Initial rows:", initial_rows)
print("Final rows:", final_rows)
print("Total rows removed:", initial_rows - final_rows)
```

Consequences of Cleaning

Original dataset --> 28,795 rows

Rows with missing rental prices were removed because price is required for the planned rental-price comparison.

The name, host_name, and license columns were removed because they were not required for the planned analysis.
